```python
from pathlib import Path

#The path of our data
data = Path("ICDAR-2019-SROIE/data")

#3 types of folders in our data folder : images , json containing real values extracted from images and box representing the position of text extracted
for folder in ["img", "box", "key"]:
    #Sort each type of file
    files = sorted((data / folder).iterdir())
    print(f"{folder:4s} -> {len(files)} files, first one: {files[0].name}")
```

```python
from IPython.display import Image, display
import json

#Choosing an image for display
receipt_id = "180"

#Show the receipt image
img_path = next((data / "img").glob(f"{receipt_id}.*"))
display(Image(filename=str(img_path), width=350))

#Show the correct answers (json format)
answers = json.loads((data / "key" / f"{receipt_id}.json").read_text())
print(json.dumps(answers, indent=2))
```

```python
import pytesseract
#Checking the version of pytesseract / tesseract that will allow us to extract text from image
print("Tesseract version:", pytesseract.get_tesseract_version())
```

```python
from pathlib import Path
from PIL import Image

data = Path("ICDAR-2019-SROIE/data")

img = Image.open(data / "img" / "180.jpg")   
text = pytesseract.image_to_string(img)    #Reading all text from the image
print(text)
```
